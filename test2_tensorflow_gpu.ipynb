```python
import tensorflow as tf

# Check if CUDA was built into the binary
print("Built with CUDA:", tf.test.is_built_with_cuda())

# List available GPUs
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    print(f"Found {len(gpus)} GPU(s):")
    for gpu in gpus:
        print(f" - {gpu}")
else:
    print("No GPU detected. Check your drivers and CUDA path.")
```
